## Уравнение:
$$4xu`` + 2u` + u = 0$$

## Начальные условия
$$u(0) = 1, u`(0) = -0.5$$

```python
import math
import matplotlib.pyplot as plt
```

```python
def teilor(x):
    return 1 - 1 / 2 * x + 1 / 24 * x ** 2 - 1 / 720 * x ** 3

def analytic(x):
    return math.cos(math.sqrt(x))

def euler(x0, y0, dy0, h, n):
    res = []
    eps = 1e-6
    x, y, dy = x0, y0, dy0
    
    for _ in range(n):
        if abs(x) > eps:
            ddy = -(2 * dy + y) / (4 * x)
        else:
            ddy = 1 / 12
            
        y += h * dy
        dy += h * ddy
        x += h
        res.append(y)
        
    return res
```

```python
x0 = 0
x_max = 10
N = 100
h = (x_max - x0) / N
X = []
Y_analitic = []
Y_teilor = []
Y_euler = euler(0, 1, -0.5, h, N)

for i in range(N):
    x = x0 + i * h
    X.append(x)
    Y_analitic.append(analytic(x))
    Y_teilor.append(teilor(x))
    
plt.plot(X, Y_analitic, label='Аналитическое решение')
plt.plot(X, Y_teilor, label='Тейлор')
plt.plot(X, Y_euler, label='Эйлер')
plt.legend()
plt.show()
```
