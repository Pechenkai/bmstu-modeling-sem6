## Уравнение:
$$1 - 2xuu` = u^3u`$$

## Начальные условия
$$u(0.5) = 0$$

```python
import math
import matplotlib.pyplot as plt
```

```python
def analytic(u):
    return math.exp(u * u) - u * u / 2 - 0.5

def pikard1(u):
    return 0.5 + u ** 2 / 2 + u ** 4 / 4

def pikard2(u):
    return 0.5 + u ** 2 / 2 + u ** 4 / 2 + u ** 6 / 12

def pikard3(u):
    return 0.5 + u ** 2 / 2 + u ** 4 / 2 + u ** 6 / 6 + u ** 8 / 48

def pikard4(u):
    return 0.5 + u ** 2 / 2 + u ** 4 / 2 + u ** 6 / 6 + u ** 8 / 24 + u ** 10 / 240
```

```python
x0 = 0
x_max = 1.5
N = 100
h = (x_max - x0) / N
X = []
Y_analytic = []
Y_first = []
Y_second = []
Y_third = []
Y_forth = []

for i in range(N):
    x = x0 + i * h
    X.append(x)
    Y_analytic.append(analytic(x))
    Y_first.append(pikard1(x))
    Y_second.append(pikard2(x))
    Y_third.append(pikard3(x))
    Y_forth.append(pikard4(x))
    
plt.plot(Y_analytic, X, label='Аналитическое')
plt.plot(Y_first, X, label='1 приближение Пикара')
plt.plot(Y_second, X, label='2 приближение Пикара')
plt.plot(Y_third, X, label='3 приближение Пикара')
plt.plot(Y_forth, X, label='4 приближение Пикара')
plt.legend()
plt.show()
```
